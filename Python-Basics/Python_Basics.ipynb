```python
# QR CODE GENERATOR
# pip install qrcode[pil]

import qrcode  # fixed the import
data = input('Enter the text or URL: ').strip()
filename = input('Enter the filename (e.g. "myqr.png"): ').strip()

qr = qrcode.QRCode(box_size=10, border=4)
qr.add_data(data)
qr.make(fit=True)  # Ensure the QR code is generated
image = qr.make_image(fill_color='black', back_color='white')
image.save(filename)

print(f'QR code saved as {filename}')
```
