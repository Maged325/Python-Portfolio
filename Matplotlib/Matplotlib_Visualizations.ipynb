### some references
#### - **MARKER** REFERENCE= 'o', '.', '*', ',', 'x', 'X'  
#### - **LINE** REFERENCE= '-', '--', ':', '-.'
#### - **COLOR** REFERENCE= 'r', 'g', 'b', 'y', 'k', 'w' 
#### - **dpi**  specifies pixels per inch. 300 is good when saving so graph has high resolution
#### - **xticks/yticks** scale mark of graph
#### - **mfc:** to set colour inside marker
#### - **mec:** to set colour of the edge marker

# **1. Line Plot**

## Graph_1

```python
x=[0,1,2,3,4]     # giving range for x-axis
y=[0,2,4,6,8]     # giving range for y-axis

# Resize the graph
plt.figure(figsize=(5,3),dpi=100)

# plt.plot(x,y,label='2x',color='blue',linewidth=5,linestyle='--',marker='.',markersize=20,markeredgecolor='red')

# scale mark of graph
plt.xticks([0,1,2,3,4])
plt.yticks([0,2,4,6,8])

plt.plot(x,y,'b^--',label='2x')
# Use shorthand notation
# fmt='[color][marker][line]'

plt.title('First Graph!',fontdict={'fontname':'Arial','fontsize':20})
plt.xlabel('X-axis')
plt.ylabel('Y-axis')

plt.legend()
# plt.savefig('graph_1.png',dpi=300)
plt.show()
```

# Graph_3: **Subplots with Shared Y-axis**
### - Create two subplots in a single figure where both plots share the same y-axis. 
### - The first plot should show the number of hours studied over a week, and the second should display the number of hours of sleep.
### - Customize the plots with different line styles, markers, and colors. Add a shared y-axis label and individual x-axis labels.

```python
# importing necessary library
import matplotlib.pyplot as plt

# Data for the subplots
days = ['Sun','Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat']
hours_studied = [2, 3, 4, 5, 6, 2, 1]
hours_slept = [7, 8, 6, 7, 7, 8, 9]

# Create a figure and a set of subplots
fig, (ax1, ax2) = plt.subplots( 2,1,sharey=True, figsize=(8, 6))         # figure has 2 rows and 1 column

# Plot Axis-1: number of hours studied
ax1.plot(days, hours_studied, color='blue', marker='o', linestyle='-', label='Hours Studied')
ax1.set_xlabel('Day of the Week')
ax1.set_ylabel('Hours')
ax1.set_title('Hours Studied Over a Week')
ax1.grid(True)
ax1.legend()

# Plot Axis-2: number of hours slept
ax2.plot(days, hours_slept, color='green', marker='s', linestyle='--', label='Hours Slept')
ax2.set_xlabel('Day of the Week')
ax2.set_title('Hours Slept Over a Week')
ax2.grid(True)
ax2.legend()

# Add a shared y-axis label
fig.supylabel('Hours')

# Adjust layout to prevent overlap
plt.tight_layout()
# Show the plots
plt.show()
```
