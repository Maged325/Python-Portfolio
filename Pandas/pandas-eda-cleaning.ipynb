## 📘 **What is a DataFrame?**
A DataFrame is a 2-dimensional labeled data structure with columns of potentially different types, similar to a spreadsheet or SQL table.

## 1. Pandas Series

```python
import pandas as pd
```

```python
groceries= pd.Series(data=[30,6,'Yes','No'],index=['eggs','apples','milk','bread'])
print(groceries)
print('--------'*3)
print('banana' in groceries)
print('--------'*3)
print( groceries.values)
print('--------'*3)
print('bread' in groceries)
print('--------'*3)
print(groceries.drop('apples'))
print('--------'*3)
print(groceries['eggs'])
print('--------'*3)
print(groceries[['milk','bread']])
```

```python
fruits=pd.Series([10,6,3],['apples','oranges','bananas'])    # by default  ( data= ,   index=  )
fruits *2    #  -2,   *2,  /2
```

```python
top_name=['high','medium','low']
score=[10,25,40]
total=pd.Series(data=score,index=top_name)

print(total)
print(type(total))
```
