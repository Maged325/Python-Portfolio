# Seaborn Data Visualization in Python

- This notebook demonstrates various visualizations using the Seaborn library, built on top of Matplotlib, for data exploration and analysis. Seaborn provides an easy-to-use interface for creating beautiful and informative statistical graphics.
- These visualizations help in exploring data distributions, relationships, trends, and patterns for better data analysis and interpretation.
- In this notebook, I use some built-in datasets from the Seaborn library, such as **Iris**, **Tips**, **Flights**, and **Titanic**.

## 📊 Visualizations Covered:

1. **Line Plot:** Analyzing trends over time or ordered data.
2. **Bar Plot:** Comparing categorical data across different groups.
3. **Histogram:** Distribution of numerical data.
4. **Scatter Plot:** Visualizing relationships between two variables.
5. **Regression Plot:** Visualizes the relationship between two variables with a fitted regression line.
6. **Linear Model (lmplot):** Plots data points with a linear regression model.
7. **Distribution Plot:** Combined histogram and KDE to visualize data distribution.
8. **Pairwise Plot:** Displaying pairwise relationships between numerical variables.
9. **Heatmap:** Visual representation of data correlation or data density.
10. **seaborn.palplot():** Displays a palette of colors as a horizontal array.
11. **Point Plot:** Visualizing the relationship between two categorical variables.
12. **Box Plot:** Displaying the distribution of data with quartiles.
13. **Residual Plot:** Displays the residuals on the vertical axis and the independent variable on the horizontal axis.


## 🎨 Some Key Parameters Used in Seaborn Visualizations

- **palette="muted"**: Defines a visually appealing and calm color palette for the plot.
- **kde=True**: Adds a smooth curve to represent the distribution of data, providing insights into the data's density.
- **annot=True**: Displays numeric values inside the heatmap cells, making the data points more readable.
- **cmap='coolwarm'**: Specifies the color palette used in the heatmap to represent the correlation values, where colors range from cool to warm tones
- **sizes:** Adjusts marker sizes based on a variable
- **hue:** Groups data by a category using different colors
- **style:** Differentiates markers or lines based on a category
- **markers:** Sets custom marker shapes
- **truncate=True/False** a boolean parameter that controls whether to shorten (truncate) long data displays or outputs.
- **Scatter_kws/line_kws** is used inside sns.lmplot() and sns.regplot() to control the appearance of the scatter/line points in your plot.

### Importing Seaborn Library

```python
import seaborn as sns
```

## -----------------------------------------------------------------------------------------------------------

# 1. Line Plot 
### to visualize trends in data over time, each data point is connected by a line, creating a smooth curve.

```python
# using built-in dataset 'flights'
flights=sns.load_dataset('flights')
sns.lineplot(x='year',y='passengers',data=flights,size='month',hue='month')
sns.set_style('white')
```

```python
# using built-in dataset 'fmri'
fmri=sns.load_dataset('fmri')
sns.lineplot(x='timepoint',y='signal',hue='event',style='region', markers=True, dashes=True, data=fmri)
sns.set_style('white')
```

## ------------------------------------------------------------------------------------------------------------

# 2. Bar Plot
### each bar represents the mean or median of the continuous variable for each category

```python
# using built-in dataset 'titanic'
titanic = sns.load_dataset("titanic")
sns.barplot(x="class", y="fare", hue="sex",palette="muted", data=titanic)
```

## ------------------------------------------------------------------------------------------------------------

# 3. Histogram 
### Data is divided into bins and the height of each bin represents the frequency or count of data points within that bin

```python
# using built-in dataset 'iris'
iris = sns.load_dataset("iris")
sns.histplot(x='petal_length',data=iris)
```

```python
# using built-in dataset 'iris' with kde to make Distribution Plot
iris = sns.load_dataset("iris")
sns.histplot(data=iris,x='petal_length',kde=True,bins=20,color='green')
```

## ------------------------------------------------------------------------------------------------------------

# 4. Scatter Plot
###  In Seaborn’s scatterplot() function, you can control the size of the markers (dots) by mapping them to a variable in your dataset using the size parameter, and controlling the range of possible sizes with sizes.

## Example 1

```python
import numpy as np
import seaborn as sns
# Generate random data using NumPy library
x = np.random.rand(40)  # 100 random numbers between 0 and 1 for x-axis
y = np.random.rand(40)  # 100 random numbers between 0 and 1 for y-axis
sns.scatterplot(x=x, y=y, marker="D")
```

## Example 2 

```python
tips=sns.load_dataset('tips')
sns.set_style('darkgrid')
sns.scatterplot(data=tips, x='total_bill', y='tip',hue='sex',style='smoker',size='size',sizes=(10,110))
```

```python
sns.scatterplot(data=tips,x='total_bill',y='tip',hue='time',style='time',markers={'Lunch':'s','Dinner':'D'})
```

## ------------------------------------------------------------------------------------------------------------

# 5. Regression Plot

```python
tips=sns.load_dataset('tips')
sns.regplot(x='total_bill',y='tip',data=tips)
```

# 6. Linear Model
### To visualize the relationship between two numerical variables.

```python
tips=sns.load_dataset('tips')
sns.lmplot(x='total_bill',y='tip',data=tips,hue='smoker',col='smoker',row='sex',truncate=False)
```

```python
sns.lmplot(x='total_bill',y='tip',data=tips,hue='smoker',col='smoker',scatter_kws={'s':100,'edgecolor':'blue'},line_kws={'lw':5,'ls':'--'})
```

## ------------------------------------------------------------------------------------------------------------

# 7. Distribution Plot

```python
# using random data from NumPy
import numpy as np
import seaborn as sns
data = np.random.normal( size=1000)  
sns.histplot(data, kde=True, color="skyblue")
```

## ------------------------------------------------------------------------------------------------------

# 8. Pairwise Relationship with pairplot()
### Some datasets contain many variables. In such cases, the relationship between each and every variable should be analyzed.

```python
tips=sns.load_dataset('tips')
sns.set_style("white")
sns.pairplot(tips)
```

## ------------------------------------------------------------------------------------------------------

# 9. HeatMap
### uses colors to depict the value of a variable in a two-dimensional space.

## Example 1

```python
# using built-in dataset 'tips'
tips = sns.load_dataset('tips')
# Select only the numeric columns for correlation
corr = tips.select_dtypes(include='number').corr()

sns.heatmap(corr,annot=True,cmap='coolwarm',linewidths=0.5)
# annot=True:  to display the numeric values inside the heatmapt cells
# cmap='coolwarm': specifies the color palette used to represent the correlation values in the heatmap.
```

## Example 2

```python
# using built-in dataset 'flights'
flights = sns.load_dataset('flights')
# Pivot the data correctly
flights_pivot = flights.pivot(index='month', columns='year', values='passengers')
# Creating heatmap
sns.heatmap(flights_pivot, cmap='Blues', annot=True,fmt='d')
```

## -------------------------------------------------------------------------------------------------------------

# 11. Point Plot
### Rather than the full bar, the value of the estimate is represented by the point at a certain height on the other axis.

```python
sns.pointplot(x="time", y="total_bill", data=tips,hue='smoker')
```

## -------------------------------------------------------------------------------------------------------------

# 12. Box Plot

```python
tips=sns.load_dataset('tips')
sns.set_style("whitegrid")
sns.boxplot(x="day", y="total_bill", data=tips)
```

## -------------------------------------------------------------------------------------------------------------

# 13. Residual Plot
### a scatterplot that displays the residuals on the vertical axis and the independent variable on the horizontal axis. 

```python
sns.residplot(x="total_bill", y="tip", data=tips)
```
