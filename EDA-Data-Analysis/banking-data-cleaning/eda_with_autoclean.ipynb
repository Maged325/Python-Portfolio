# 🧹 **Data Cleaning with AutoClean | First Look & Quick Cleanup**

## 🔍 **What This Notebook Does**

- Loads a raw dataset
- Shows basic info: shape, missing values, datatypes, and sample records
- Automatically cleans the data using the `AutoClean` library
- Shows how the dataset looks after cleaning

Use this notebook as a template for any CSV dataset you're working with in your data analysis projects!

```python
# Step 1: Import libraries
import pandas as pd
from AutoClean import AutoClean
import mitosheet
```

```python
# Step 2: Load dataset
file_path = (r'C:\Users\Maged\Desktop\Banking_Analytics_Dataset.csv')
df = pd.read_csv(file_path)
```

```python
# Step 3: Explore raw dataset
print("📌 Original Dataset Summary\n")
print(f"Shape: {df.shape}")
print(f"Duplicate rows: {df.duplicated().sum()}")
print("\nMissing values:\n", df.isnull().sum())
print("\nData types:\n", df.dtypes)
print("\n🔍 Sample records (first 5 rows):\n", df.head())
```

```python
# Step 4: Run AutoClean
cleaner = AutoClean(df)
cleaned_df = cleaner.output
```

### ⚖️ **Comparison: Before vs After AutoClean**

Below we compare key dataset properties before and after cleaning:
- Number of rows
- Column names
- Null value counts
- Data types

```python
# ✅ Step 5: Compare before vs after
print("\n⚖️ Comparison Summary\n")
print("Rows before:", len(df))
print("Rows after :", len(cleaned_df))
print("\nOriginal columns:", df.columns.tolist())
print("Cleaned columns :", cleaned_df.columns.tolist())
print("\nNulls before:\n", df.isnull().sum())
print("Nulls after:\n", cleaned_df.isnull().sum())
print("\nData types before:\n", df.dtypes)
print("\nData types after:\n", cleaned_df.dtypes)
```

```python
# Step 6: Preview cleaned data
print("\n👀 Sample cleaned data:\n")
print(cleaned_df.head())
```
