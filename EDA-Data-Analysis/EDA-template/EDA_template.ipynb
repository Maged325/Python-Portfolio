## 🧹**5. Data Filtering & Sorting**

#### 1. **Using 'loc' to locate specific rows**

```python
print(df.loc[[0,1]])    # to get data of only row 1 & 2
print(df.loc[0])        # to get data on only row 1
```

#### 2. **To find specific data in a column**

```python
df[df['First_Name'].str.contains('Peter')]        # Example_1
df[df['CustomerID'] <= 1010]                      # Example_2
df[(df['Paying Customer']=='No') & (df['Do_Not_Contact']=='Yes')]        # Example_3: with two columns
```

#### 3. **To sort data**

```python
df[df['Rank']<10].sort_values(by = 'Rank')                      # Example_1: in ascending order
df[df['Rank']<10].sort_values(by = 'Rank',ascending=False)      # Example_2: in descending order
df[df['Rank']<10].sort_values(by = ['Country','Capital'])       # Example_3: by two columns
```

#### 4. **Calling out Columns**

```python
df.Address                  # calling out a column
df['Paying Customer']       # calling out a column with SPACES
df[['First_Name','Last_Name','Address']]      # calling out multiple columns
```

#### 5. **Unique Values**

```python
df.Column_X.unique()          # get unique values   for 'Column_X'
df['Column X'].unique()       # get unique values for column with spaces 'Column X'
```

#### 6. **DateTime Column**

```python
df["year"] = df["date_column"].dt.year    # extracting year from date_column
```

```python
df["month"] = df["date_column"].dt.month    # extracting month from date_column
```

```python
df["day"] = df["date_column"].dt.day     # extracting day from date_column
```

```python
df['hour'] = df['date_column'].dt.hour   # extracting hour from date_column
```

```python
 # To convert the date_column from string/object type to datetime format
df["date_column"] = pd.to_datetime(df["date_column"])    
```

#### 7. **Grouping By**

```python
# Grouping by unique values in Column_X, then sums all numeric columns within each group.
df.groupby("column_X").sum()    # min(), max(), mean()
```

```python
# Group data by Column_X and count non-values for only Column_Y, then Plot bar chart
df.groupby('Column_X').count()['Column_Y']
df.groupby('Column_X').count()['Column_Y'].plot(kind='bar')
plt.title('Title_Name')    # plot title
```
