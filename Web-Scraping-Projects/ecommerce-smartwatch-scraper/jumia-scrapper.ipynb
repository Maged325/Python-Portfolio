## 📘This project performs web scraping on e-commerce Egypt website (**Jumia**) to search for smartwatches. It collects key product details such as:

### It collects key product details such as:
- Product Name
- Price
- Discount
- Rating
- Product Link

### Websites included:
- Jumia Egypt  

### The scraped data is exported to a CSV file saved in your current working directory.

---

### **📚 Importing Libraries**

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

### **🔍Scraping Function**

```python
def jumia(link):
    global products_list  
    try:
        # getting the elements from the page
        page = requests.get(link)
        soup = BeautifulSoup(page.text, 'html.parser')
        products_list = soup.find_all('article', {'class': 'prd'})
        print("Page loaded and parsed successfully.")
        return products_list
    except Exception as e:
        print(f"Error occurred during Jumia scraping: \n {e}")   
        return []
```

```python
# Calling the function
jumia('https://www.jumia.com.eg/catalog/?q=smart+watches')    
```

### **📋 Extracting Product Details**
#### Extracts product details (name, price, rating, discount, and link) from the scraped product list and stores them in a structured list.

```python
products_list = jumia('https://www.jumia.com.eg/catalog/?q=smart+watches')    
products_details = []

for product in products_list:
    try:
        products_name = product.find('h3', {'class': 'name'}).text.strip()
    except:
        products_name = 'No Name found'
        
    try:
        products_price = product.find('div', {'class': 'prc'}).text.strip()
    except:
        products_price = 'No Price found'

    try:
        products_rate = product.find('div', {'class': 'stars'}).text.strip()
    except:
        products_rate = 'No Rate found'

    try:
        products_discount = product.find('div', {'class': 'bdg'}).text.strip()
    except:
        products_discount = 'No Discount found'

    try:
        product_link = f"https://www.jumia.com.eg{product.find('a', {'class': 'core'}).get('href')}"
    except:
        product_link = 'No Link found'

    products_details.append({
        'Website': 'JUMIA',
        'PRODUCT NAME': products_name,
        'PRODUCT PRICE': products_price,
        'PRODUCT DISCOUNT': products_discount,
        'PRODUCT RATE': products_rate,
        'PRODUCT LINK': product_link
    })

print(f"Extracted {len(products_details)} products.")
```

### **📢 Preview Scraped Products**

```python
# Viewing the first 5 products
for i, product in enumerate(products_details[:5], 1):   
    print("\n----- Product no.", i, "-----")
    for key, value in product.items():
        print(key + ":", value)
```

### **💾 Export to CSV**

```python
df = pd.DataFrame(products_details)
df.to_csv('jumia_smart_watches.csv', index=False, encoding='utf-8-sig')
print("✅ Data successfully exported to 'jumia_smart_watches.csv'. The file is saved in your working directory.")
```
