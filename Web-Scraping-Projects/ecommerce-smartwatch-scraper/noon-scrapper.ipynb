```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
import time
import pandas as pd
```

### ⚙️**Configure Selenium WebDriver**

```python
options = Options()
options.add_argument("--start-maximized")

driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=options)
driver.get("https://www.noon.com/egypt-en/search/?q=smart%20watches")

# Wait for page to load (adjust time if needed)
time.sleep(5)
```

### 🔍**Locate Products on the Page**

```python
products = driver.find_elements(By.CSS_SELECTOR, 'div.ProductBoxLinkHandler_linkWrapper__b0qZ9')  # this may need to be updated
print(f"Found {len(products)} products on the page.")
```

### **📋 Extracting Product Details**
#### Extracts product details (name, price, rating, discount, and link) from the scraped product list and stores them in a structured list.

```python
from selenium.webdriver.common.by import By

product_details = []

for product in products:
    try:
        name = product.find_element(By.CSS_SELECTOR, 'div.Text_text__eAaVj.Text_medium__jZEox.Text_left__GJB9G.Text_textTransformCapitalize__OkA1S.Text_ellipsis__QqS_z.Text_fontBodySmall__scGmT').text
    except:
        name = 'No Name found'

    try:
        price = product.find_element(By.CSS_SELECTOR, 'div.Price_currentPrice__mG8cv').text
    except:
        price = 'No Price found'

    try:
        link = product.find_element(By.TAG_NAME, 'a').get_attribute('href')
    except:
        link = 'No Link found'

    try:
        discount=product.find_element(By.CSS_SELECTOR,'div.Price_preReductionRow__yBUIC').text
    except:
        discount= "No Discount found"
    try:
        rate=product.find_element(By.CSS_SELECTOR,'div.RatingPreviewStar_textCtr__sfsJG').text
    except:
        discount= "No Rating found"   

    product_details.append({
        'Website': 'NOON',
        'PRODUCT NAME': name,
        'PRODUCT PRICE': price,
        'PRODUCT DISCOUNT':discount,
        'PRODUCT RATE':rate,
        'PRODUCT LINK': link
    })

print(f"Extracted {len(product_details)} products.")
```

### **📢 Preview Scraped Products**

```python
# 📋 Viewing the first 5 products
for i, product in enumerate(products_details[:5], 1):   
    print(f"\n----- Product-no. {i} -----")
    for key, value in product.items():
        print(f"{key}: {value}")
```

### **💾Export to CSV**

```python
import pandas as pd

# Convert to DataFrame
df = pd.DataFrame(product_details)

# Export to CSV
df.to_csv('noon_smart_watches.csv', index=False, encoding='utf-8-sig')

print("✅ Data successfully exported to 'noon_smart_watches.csv'. The file is saved in your working directory.")
```
